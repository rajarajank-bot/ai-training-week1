```python
import pandas as pd

# IBM HR Analytics Dataset
url = "https://raw.githubusercontent.com/ibm-developer-skills-network/all-datasets/master/hr_analytics.csv"
df = pd.read_csv(url)

print("--- 1. Shape of Data ---")
print(df.shape)

print("\n--- 2. Data Info ---")
df.info()

print("\n--- 3. Statistical Summary ---")
print(df.describe())

print("\n--- 4. Missing Values ---")
print(df.isna().sum())

print("\n--- 5. Duplicate Rows ---")
print(df.duplicated().sum())
```
