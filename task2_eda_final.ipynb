```python
import pandas as pd
import numpy as np

# 1. Dataset Load 
df = pd.read_csv(r"C:\Users\Admin\clean_data.csv")

# Requirement 1: Summary statistics (mean, median, min, max, std)
print("=== 1. Summary Statistics (Salary) ===")
summary_stats = df[['salary']].agg(['mean', 'median', 'min', 'max', 'std']).round(2).T
display(summary_stats)

# Requirement 2: groupby().agg() group comparison
print("\n=== 2. Department-wise Salary Breakdown ===")
dept_group = df.groupby('department')['salary'].agg(['count', 'mean', 'median', 'std']).round(2)
display(dept_group)

print("\n=== City-wise Salary Breakdown ===")
city_group = df.groupby('city')['salary'].agg(['count', 'mean', 'median']).round(2)
display(city_group)

# Requirement 3: Correlations with df.corr()
print("\n=== 3. Correlation Matrix ===")
display(df[['id', 'salary']].corr().round(4))

# Requirement 4: Business questions answered with numbers
print("\n=== 4. Business Questions & Answers ===")
top_dept = df.groupby('department')['salary'].mean().idxmax()
top_dept_sal = df.groupby('department')['salary'].mean().max()
print(f"Q1: Highest average salary department: {top_dept} (₹{top_dept_sal:,.2f})")

avg_sal = df['salary'].mean()
med_sal = df['salary'].median()
print(f"Q2: Organization Benchmark: Mean = ₹{avg_sal:,.2f} | Median = ₹{med_sal:,.2f}")

top_city = df['city'].value_counts().idxmax()
emp_count = df['city'].value_counts().max()
print(f"Q3: City with highest employee count: {top_city} ({emp_count} employees)")

# Requirement 5: Merge department budget table (pd.merge)
budget_df = pd.DataFrame({
    'department': ['Sales', 'Hr', 'It', 'Finance'],
    'budget': [300000, 200000, 800000, 500000]
})
merged_df = pd.merge(df, budget_df, on='department', how='left')
print("\n=== 5. Merged Table (Employee Data + Budget) ===")
display(merged_df[['id', 'name', 'department', 'salary', 'budget']])

# Bonus Challenge: Pivot table with 2 categorical columns
print("\n=== 6. Bonus Challenge: Pivot Table (Department vs City) ===")
pivot_tbl = pd.pivot_table(
    df,
    values='salary',
    index='department',
    columns='city',
    aggfunc='mean',
    fill_value=0
).round(2)
display(pivot_tbl)
```
