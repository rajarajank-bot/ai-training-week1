```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Reports folder create 
os.makedirs("reports/figures", exist_ok=True)

# Cleaned data load 
df = pd.read_csv(r"C:\Users\Admin\clean_data.csv")

# Consistent colour scheme
sns.set_theme(style="whitegrid", palette="Blues_r")

# 1. Bar Chart: Department vs Mean Salary
plt.figure(figsize=(7, 4))
dept_sal = df.groupby('department')['salary'].mean().reset_index()
sns.barplot(data=dept_sal, x='department', y='salary')
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/1_bar_dept_salary.png")
plt.show()
print("Insight: Finance department records the highest average salary among all departments.")

# 2. Histogram: Salary Distribution
plt.figure(figsize=(7, 4))
sns.histplot(df['salary'], kde=True, bins=5, color="#1f77b4")
plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary (INR)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("reports/figures/2_hist_salary.png")
plt.show()
print("Insight: Majority of employee salaries fall in the 40,000 to 75,000 range.")

# 3. Scatter Plot: ID vs Salary
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='id', y='salary', hue='department', s=100)
plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary (INR)")
plt.legend(title="Department")
plt.tight_layout()
plt.savefig("reports/figures/3_scatter_id_salary.png")
plt.show()
print("Insight: Higher salary packages are concentrated in specific departments across employee IDs.")

# 4. Line Chart: Sorted Salary Progression
plt.figure(figsize=(7, 4))
sorted_df = df.sort_values('salary').reset_index(drop=True)
plt.plot(sorted_df.index, sorted_df['salary'], marker='o', linestyle='-', color='#2b5c8f')
plt.title("Employee Salary Trend (Ascending)")
plt.xlabel("Employee Rank")
plt.ylabel("Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/4_line_salary_trend.png")
plt.show()
print("Insight: Steady progression observed from entry-level pay to senior-level compensation.")

# 5. Boxplot: Salary by City
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='city', y='salary')
plt.title("Salary Variation Across Cities")
plt.xlabel("City")
plt.ylabel("Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/5_box_city_salary.png")
plt.show()
print("Insight: Mumbai and Bangalore show higher median pay compared to Chennai.")
```
